```python
from ucimlrepo import fetch_ucirepo 
import pandas as pd  
from sklearn.model_selection import train_test_split
```

# Bank

```python

# fetch dataset 
bank_marketing = fetch_ucirepo(id=222) 
  
# data (as pandas dataframes) 
X = bank_marketing.data.features 
y = bank_marketing.data.targets 

print("total length",len(X))

merged_df = pd.concat([X, y], axis=1)
merged_df.isna().sum()
```

```python
pd.unique(merged_df["previous"])
```

```python
merged_df = merged_df.drop(columns=['poutcome','contact','pdays'])
merged_df=merged_df.dropna()
# merged_df['y'] = merged_df['y'].map({'no': 0, 'yes': 1})

merged_df.head()
```

```python
merged_df['age_group'] = merged_df['age'].apply(lambda x: 1 if 25 < x < 65 else 0)

merged_df.head()
```

```python
age_group_column = merged_df.pop('age_group')  # Remove the column
merged_df.insert(8, 'age_group', age_group_column)  # Insert it at position 9
merged_df.head()
```

```python
columns = list(merged_df.columns)
columns.remove("age_group") 
columns.remove("marital") 
columns.insert(8, "marital") 
columns.insert(9, "age_group") 
merged_df = merged_df[columns]
merged_df.head()
```

```python
# merged_df['job'] = merged_df['job'].str.strip()
# merged_df['marital'] = merged_df['marital'].str.strip()
# merged_df['education'] = merged_df['education'].str.strip()
# merged_df['default'] = merged_df['default'].str.strip()
# merged_df['housing'] = merged_df['housing'].str.strip()
# merged_df['loan'] = merged_df['loan'].str.strip()
# merged_df['month'] = merged_df['month'].str.strip()
# merged_df['y'] = merged_df['y'].str.strip()
# merged_df.tail()
```

```python
len(merged_df)
```

```python
merged_df['martial_ageGroup'] = merged_df['marital'] + '_' + merged_df['age_group'].astype(str)

# Step 2: Stratified Sampling for df1 (10,000 data points)
df1_sample, remaining_data = train_test_split(
    merged_df, test_size=(1 - 6000 / len(merged_df)), stratify=merged_df['martial_ageGroup'], random_state=42
)
df1_sample = df1_sample.drop(columns=['martial_ageGroup'])
df1_sample = df1_sample.reset_index(drop=True)
df1_sample
```

```python
len(remaining_data)
```

```python
df2_sample, remaining_data = train_test_split(
    remaining_data, test_size=(1 - 10000 / len(remaining_data)), stratify=remaining_data['martial_ageGroup'], random_state=42
)
df2_sample = df2_sample.drop(columns=['martial_ageGroup'])

df2_sample = df2_sample.reset_index(drop=True)

df2_sample
```

```python
len(df2_sample),len(df1_sample)
```

```python
len(remaining_data)
```

```python
dfs_2000 = []

while len(remaining_data) >= 2000:
    sample_2000, remaining_data = train_test_split(
        remaining_data, test_size=(1 - 2000 / len(remaining_data)), stratify=remaining_data['martial_ageGroup'], random_state=42
    )
    dfs_2000.append(sample_2000)

# Handle remaining data points (if any) as the last DataFrame
if len(remaining_data) > 0:
    dfs_2000.append(remaining_data)

dfs_2000 = [df.drop(columns=['martial_ageGroup']) for df in dfs_2000]
dfs_2000 = [df.reset_index(drop=True) for df in dfs_2000]
```
