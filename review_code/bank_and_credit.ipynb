```python
from ucimlrepo import fetch_ucirepo 
import pandas as pd  
from sklearn.model_selection import train_test_split
```

# Credit Data

```python

# Fetch dataset
default_of_credit_card_clients = fetch_ucirepo(id=350)

# Extract features (X) and target (y)
X = default_of_credit_card_clients.data.features
y = default_of_credit_card_clients.data.targets

# Metadata inspection for column names
# print("Metadata:")
# print(default_of_credit_card_clients.metadata)

# Manually assign meaningful column names based on dataset documentation
column_names = [
    "LIMIT_BAL", "SEX", "EDUCATION", "MARRIAGE", "AGE",
    "PAY_1", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6",
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3", "BILL_AMT4", "BILL_AMT5", "BILL_AMT6",
    "PAY_AMT1", "PAY_AMT2", "PAY_AMT3", "PAY_AMT4", "PAY_AMT5", "PAY_AMT6"
]

# Rename the columns in X
X.columns = column_names

```

```python
merged_df = pd.concat([X, y], axis=1)
```

```python
merged_df['age_group'] = merged_df['AGE'].apply(lambda x: 1 if 25 < x < 65 else 0)
age_group_column = merged_df.pop('age_group')  # Remove the column
merged_df.insert(8, 'age_group', age_group_column)  # Insert it at position 9

merged_df.head()
```

```python
merged_df['sex_ageGroup'] = merged_df['SEX'].astype(str) + '_' + merged_df['age_group'].astype(str)

# Step 2: Stratified Sampling for df1 (10,000 data points)
df1_sample, remaining_data = train_test_split(
    merged_df, test_size=(1 - 6000 / len(merged_df)), stratify=merged_df['sex_ageGroup'], random_state=42
)
df1_sample = df1_sample.drop(columns=['sex_ageGroup'])
df1_sample = df1_sample.reset_index(drop=True)
df1_sample
```

```python
len(remaining_data)
```

```python
df2_sample, remaining_data = train_test_split(
    remaining_data, test_size=(1 - 10000 / len(remaining_data)), stratify=remaining_data['sex_ageGroup'], random_state=42
)
df2_sample = df2_sample.drop(columns=['sex_ageGroup'])

df2_sample = df2_sample.reset_index(drop=True)

df2_sample
```

```python
len(df2_sample),len(df1_sample)
```

```python
len(remaining_data)
```

```python
dfs_2000 = []

while len(remaining_data) >= 2000:
    sample_2000, remaining_data = train_test_split(
        remaining_data, test_size=(1 - 2000 / len(remaining_data)), stratify=remaining_data['sex_ageGroup'], random_state=42
    )
    dfs_2000.append(sample_2000)

# Handle remaining data points (if any) as the last DataFrame
if len(remaining_data) > 0:
    dfs_2000.append(remaining_data)

dfs_2000 = [df.drop(columns=['sex_ageGroup']) for df in dfs_2000]
dfs_2000 = [df.reset_index(drop=True) for df in dfs_2000]
```
